## Criando tabelas dimensão

```python
from pyspark.sql.functions import col
from pyspark.sql.types import IntegerType, DoubleType, TimestampType

# Caminhos
bronze_base = "/mnt/datalakef085704a8c687020/bronze"
silver_base = "/mnt/datalakef085704a8c687020/silver"

# ---------------------------
# 1. Dimensão Cliente
# ---------------------------
df_cliente = spark.read.format("delta").load(f"{bronze_base}/customer")
df_cliente_clean = df_cliente.select(
    col("customer_id").cast(IntegerType()),
    col("full_name"),
    col("email"),
    col("phone_number"),
    col("registration_date").cast(TimestampType())
)
df_cliente_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_cliente")

# ---------------------------
# 2. Dimensão Endereço
# ---------------------------
df_address = spark.read.format("delta").load(f"{bronze_base}/address")
df_address_clean = df_address.select(
    col("address_id").cast(IntegerType()),
    col("customer_id").cast(IntegerType()),
    col("zip_code"),
    col("street_address"),
    col("city"),
    col("state"),
    col("address_type")
)
df_address_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_endereco")

# ---------------------------
# 3. Dimensão Categoria
# ---------------------------
df_categoria = spark.read.format("delta").load(f"{bronze_base}/category")
df_categoria_clean = df_categoria.select(
    col("category_id").cast(IntegerType()),
    col("category_name")
)
df_categoria_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_categoria")

# ---------------------------
# 4. Dimensão Fornecedor
# ---------------------------
df_fornecedor = spark.read.format("delta").load(f"{bronze_base}/supplier")
df_fornecedor_clean = df_fornecedor.select(
    col("supplier_id").cast(IntegerType()),
    col("supplier_name"),
    col("cnpj"),
    col("contact_email")
)
df_fornecedor_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_fornecedor")

# ---------------------------
# 5. Dimensão Produto
# ---------------------------
df_produto = spark.read.format("delta").load(f"{bronze_base}/product")
df_produto_clean = df_produto.select(
    col("product_id").cast(IntegerType()),
    col("category_id").cast(IntegerType()),
    col("supplier_id").cast(IntegerType()),
    col("product_name"),
    col("description"),
    col("price").cast(DoubleType()),
    col("stock_quantity").cast(IntegerType())
)
df_produto_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_produto")

# ---------------------------
# 6. Dimensão Promoção
# ---------------------------
df_promocao = spark.read.format("delta").load(f"{bronze_base}/promotion")
df_promocao_clean = df_promocao.select(
    col("promotion_id").cast(IntegerType()),
    col("promo_code"),
    col("description"),
    col("discount_percentage").cast(DoubleType()),
    col("expiration_date").cast(TimestampType())
)
df_promocao_clean.write.format("delta").mode("overwrite").save(f"{silver_base}/dim_promocao")
```
